```python
import pandas as pd

daily = pd.concat([pd.read_csv("activity_p1.csv.csv"),
                   pd.read_csv("activity_p2.csv.csv")], ignore_index=True)
steps = pd.concat([pd.read_csv("steps_p1.csv"),
                   pd.read_csv("steps_p2.csv")], ignore_index=True)

daily.info()
steps.info()
print("Users (daily):", daily["Id"].nunique())
print("Users (hourly):", steps["Id"].nunique())
```

```python
# Daily
daily["ActivityDate"] = pd.to_datetime(daily["ActivityDate"], format="%m/%d/%Y")
before = len(daily)
daily = daily.drop_duplicates(subset=["Id", "ActivityDate"])
print("Daily duplicates removed:", before - len(daily))

daily["DayOfWeek"] = daily["ActivityDate"].dt.day_name()
daily["IsWeekend"] = daily["ActivityDate"].dt.dayofweek >= 5
daily["LikelyNotWorn"] = daily["TotalSteps"] == 0

print("Date range:", daily["ActivityDate"].min(), "to", daily["ActivityDate"].max())
print("Zero-step days:", daily["LikelyNotWorn"].sum())

# Hourly
steps["ActivityHour"] = pd.to_datetime(steps["ActivityHour"], format="%m/%d/%Y %I:%M:%S %p")
before = len(steps)
steps = steps.drop_duplicates(subset=["Id", "ActivityHour"])
print("Hourly duplicates removed:", before - len(steps))

steps["Hour"] = steps["ActivityHour"].dt.hour
steps["Date"] = steps["ActivityHour"].dt.date
```

```python
daily = daily.copy()
steps = steps.copy()
daily["ActivityDate"] = daily["ActivityDate"].dt.strftime("%Y-%m-%d")
steps["ActivityHour"] = steps["ActivityHour"].dt.strftime("%Y-%m-%d %H:%M:%S")
steps["Date"] = steps["Date"].astype(str)
daily["LikelyNotWorn"] = daily["LikelyNotWorn"].astype(int)
daily["IsWeekend"] = daily["IsWeekend"].astype(int)
```
